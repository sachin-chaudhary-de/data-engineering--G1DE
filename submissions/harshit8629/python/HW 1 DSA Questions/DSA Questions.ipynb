# Q24 — Container With Most Water

**Difficulty:** Medium
**Topics:** Array, two pointers

You are given an integer array `height`.

Each element represents the height of a vertical line at that index.

Choose two lines that, together with the x-axis, form a container that holds the most water.

Return the maximum amount of water that can be contained.

### Example

```python
height = [1, 8, 6, 2, 5, 4, 8, 3, 7]
```

Expected:

```python
49
```

The best pair is formed by heights `8` and `7`.

The area is:

```text
min(8, 7) * distance
= 7 * 7
= 49
```

### Function

```python
def max_area(height: list) -> int:
    pass
```

### Requirements

* Do not use nested loops.
* Aim for `O(n)` time.
* Aim for `O(1)` extra space.

### Think carefully

For two selected indices `left` and `right`:

```text
width = right - left
height = min(height[left], height[right])
area = width * height
```

Your challenge is determining **how to move the pointers**.

````python
def max_area(height: list) -> int:
    left = 0
    right = len(height) - 1
    max_water = 0
    while left < right:
        width = right - left
        h = min(height[left], height[right])
        area = width * h
        max_water = max(max_water, area)
        if height[left] < height[right]:
            left += 1
        else:
            right -= 1
    return max_water
height = [1, 8, 6, 2, 5, 4, 8, 3, 7]
print(max_area(height))
````

### Requirements

* Non-anagrams must be in different groups.
* Aim for better than `O(n²)` overall behavior.

### Important

Ask yourself:

> What can represent an anagram consistently so that I can use it as a dictionary key?

````python
def group_anagrams(strs: list) -> list:
  dict2={}
  for i in range(len(strs)):
    dict2[strs[i]]="".join(sorted(strs[i]))
  list1=[]
  visited = set()
  for i in range(len(strs)):
    if strs[i] in visited:
            continue
    list2=[]
    for j in range(len(strs)):
      if dict2[strs[i]]==dict2[strs[j]]:
        list2.append(strs[j])
        visited.add(strs[j])
    list1.append(list2)
  return list1
strs = ["eat", "tea", "tan", "ate", "nat", "bat"]
group_anagrams(strs)
````

# Q29 — Subarray Sum Equals K

**Difficulty:** Medium
**Topics:** Prefix sum, hashmap

Given an integer array `nums` and an integer `k`, return the total number of **continuous subarrays** whose sum equals `k`.

### Example

```python
nums = [1, 1, 1]
k = 2
```

Expected:

```python
2
```

The valid subarrays are:

```text
[1, 1]   # positions 0–1
[1, 1]   # positions 1–2
```

Another:

```python
nums = [1, 2, 3]
k = 3
```

Expected:

```python
2
```

because:

```text
[1, 2]
[3]
```

### Function

```python
def subarray_sum(nums: list, k: int) -> int:
    pass
```

### Requirements

Aim for:

```text
Time  : O(n)
Space : O(n)
```

### Important

Do not use nested loops.

The interesting part is not finding one subarray; it is efficiently counting **all** valid subarrays.

````python
def subarray_sum(nums: list, k: int) -> int:
    prefix_sum = 0
    count = 0
    hashmap = {0: 1}
    for num in nums:
        prefix_sum += num
        diff = prefix_sum - k
        if diff in hashmap:
            count += hashmap[diff]
        hashmap[prefix_sum] = hashmap.get(prefix_sum, 0) + 1
    return count
nums = [1, 1, 1]
k = 2
subarray_sum(nums, k)
````

**Difficulty:** Easy

A pivot index is an index where:

```text

=
sum of elements strictly to the right
```

The pivot element itself is not included in either sum.

Return `-1` if no pivot exists.

### Example

```python

```

Expected:

```python

```

Because at index `3`:



### Requirements

Aim for:

```text

Space : O(1)
```

````python
def pivot_index(nums):
    for i in range(len(nums)):
        left = sum(nums[:i])
        right = sum(nums[i+1:])
        if left == right:
            return i
    return -1

print(pivot_index([1, 7, 3, 6, 5, 6]))
print(pivot_index([1, 2, 3]))
````

### Q40 — 3Sum

Given an integer array `nums`, return all unique triplets:

```text

```

```text

```


Expected:

```python
    [-1, 0, 1]
```

The order of the triplets does not matter, and duplicate triplets must not appear.

```python

    pass
```

Aim for:

```text
O(n²) time
```

rather than `O(n³)`.

````python
def three_sum(nums):
    result = []
    n = len(nums)
    for i in range(n):
        for j in range(i + 1, n):
            for k in range(j + 1, n):
                if nums[i] + nums[j] + nums[k] == 0:
                    triplet = sorted([nums[i], nums[j], nums[k]])
                    if triplet not in result:
                        result.append(triplet)
    return result

print(three_sum([-1, 0, 1, 2, -1, -4]))
````

### Q45 — Merge Intervals

Given an array of intervals:

```python

```

merge all overlapping intervals.

Expected:

```python
[[1,6],[8,10],[15,18]]
```

Another:

```python

```

Expected:

```python
[[1,5]]
```

```python

    pass
```

Think carefully about what ordering, if any, would make this problem easier.

````python
def merge_intervals(intervals):
    if not intervals:
        return []
    intervals = sorted(intervals)
    merged = [list(intervals[0])]
    for start, end in intervals[1:]:
        last = merged[-1]
        if start <= last[1]:
            if end > last[1]:
                last[1] = end
        else:
            merged.append([start, end])
    return merged

print(merge_intervals([[1,3],[2,6],[8,10],[15,18]]))
print(merge_intervals([[1,4],[4,5]]))
````
