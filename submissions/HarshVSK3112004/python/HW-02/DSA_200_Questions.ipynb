### Q1 — Find the Largest Number
You are given a list of integers representing daily transaction values.
Input:  
* transactions = [1200, 450, 3200, 1800, 2750]

Requirements:  
* Do not use __max()__.
* Do not sort the array.
* The function should work for negative numbers as well.

````python
## Approach -1
# Step-1 Start by iterating items of array using for loop
# Step-2 now, compare items, easy start is fix one number and compare other numbers with this. 
# if the other number is small do nothing but if it is bigger than the fixed number than make that bigger and go on.

def largest_num(arr : list)-> int :
    largest = arr[0] # fixed the number and set to largest
    for item in arr: 
        if item > largest:
            largest = item
    return largest

# Driver code
transactions = [1200, 450, 3200, 1800, 2750]
largest_num(transactions)
````

### Q2 — Minimum Value  
You are processing CPU execution times for a batch job:  
Input:  
execution_times = [42, 18, 35, 11, 27, 19]  
Return the smallest execution time.

Constraints:

    Do not use min().
    Do not sort.
    Time complexity should be O(n).
    Space complexity should be O(1).

````python
# Approach
# Again same approacha as above. set the first element as minimum execution time and then compare

def min_executin_time(arr : list) -> int:

    min_time = arr[0]
    for item in arr[1:]:
        if item < min_time:
            min_time = item

    return min_time
# Drive Code
execution_times = [42, 18, 35, 11, 27, 19] 
min_executin_time(execution_times)
````

Slicing creates a new list. --> for item in arr[1:] uses O(n) extra space because of the slice.

for i in range(1, len(arr)):  
    if arr[i] < min_time:

### Q4 — Count Positive, Negative, and Zero
You receive a batch of numeric values:  
Input  
    values = [10, -5, 0, 7, -3, 0, 12, -8] 

Rules : Use a single traversal of the input. No built-in function

You need to return how many values are:  
* positive
* negative
* zero

Expected output
{
    "positive": 3,
    "negative": 3,
    "zero": 2
}


````python
## Approach
# Here we need to count 3 types of things. +, - and 0.
# # counters needed.
# Traverse array in single pass and increase the counter when an item satisfies certain type of number/condition
# since solution should be packed in dict, need dict with 3 keys

def count_negat_posit_zero(arr:list)-> int :

    counter = {'positive':0,
               'negative':0,
               'zeros':0}

    for item in arr:
        if item > 0:
            counter['positive'] += 1
        elif item < 0:
            counter['negative'] += 1
        else:
            counter['zeros'] += 1
    return counter

# Driver Code
values = [10, -5, 0, 7, -3, 0, 12, -8] 
count_negat_posit_zero(values)

````

### Q8 — Find the First Duplicate
You are receiving product IDs from an order pipeline:  
Input : product_ids = [101, 203, 405, 203, 501, 101]

Return 
* the first product ID that appears more than once.


Requirements  
Traverse from left to right.  
Return immediately when you encounter the first duplicate.  
Return None if there are no duplicates.  
Aim for O(n) time.  
You may use an appropriate Python data structure.  

````python
# Approach
## It looks extension to Q7.
# I will be counting the occurences of items as traverse and when i get second occurence of any item, i will exit and return. I won't be going till end.
# One way is use list. Iterate over arr and check if it exists in unique_list. If exists return otherwise add in unique list.
# It is similar to one of the above question where we just check first occurence.

def first_dup(arr:list) -> int|None:
    unique_list = []
    for item in arr:
        if item in unique_list:
            return item
        else:
            unique_list.append(item)
    return None

product_ids = [101, 203, 405, 203, 501, 101]

first_dup(product_ids)

# but here, time complexity is O(n2), because of extra list in if conditional.

# We can use a data structure which gives fast lookup, that is set. we need only first so we are not using dict.
# Because we do not need to count all items occurences or duplicates, only first required.

def first_duplicate(arr: list) -> int|None:
    seen = set()
    for item in arr:
        if item in seen:
            return item
        else:
            seen.add(item)
    return None

first_duplicate(product_ids)
````

### Q9 — Find the Most Frequent Product
Given: product_ids = [101, 203, 101, 405, 203, 101, 501, 203]

Return the product ID that appears most frequently.

Constraint ; If multiple products have the same highest frequency, return the one that appears first in the input.

````python
# Approach - 
## Hahahahaha, flow made this problem easy.
# Now i need to count occurences of all items and then only I can tell which item has high frequency.
# Here, to count all item occurences, i need dict definitely. And one data structure to check exists or not.
# I am not using list as complexity will increase. I will use set.
## Iterate, check if already in freq dict, if exists then increase by 1, not exists then initialise with 1

def most_frequent_product(arr: list) ->dict:
    freq = {} # empty dictionary to store counts
    for item in arr:
        if item in freq:
            freq[item] += 1
        else:
            freq[item] = 1
    return freq

product_ids = [101, 203, 101, 405, 203, 101, 501, 203]
most_frequent_product(product_ids)

# But with this, I can count only, can't select, nost frequent and that to first one.

# More condense way to do above thing
# Iterate--> Check if it is appeared-->If appeared, increase count of that item using freq[item]= freq.get(item, 0) + 1

def most_freq_prod(arr:list) -> int:
    freq = {}
    for item in arr:
        freq[item] = freq.get(item, 0) + 1

    best_count = 0
    best_prod = None

    for key, value in freq.items():
        if value > best_count:
            best_count = value
            best_prod = key
    return best_prod

most_freq_prod(product_ids)
````

Return missing id

````python
# Approach 

def missing_id(arr: list, n: int) -> int:
    expected = n*(n+1)/2
    actual = 0
    for item in arr:
        actual += item
    return int(expected - actual)

transaction_ids = [1, 2, 4, 5, 6] 
n = 6
missing_id(transaction_ids, n)
````

You receive transaction IDs from two systems:

system_a = [101, 102, 103, 105, 108]

Return the transaction IDs that are present in both systems.  
Requirements:  
* Each ID should appear only once in the output.  
* Order does not matter.  
* Aim for O(n + m) time.

````python
# Approach
#  Use 2 sets with intersection to get common items.
# Pythonic code favors methods over operators when you are dealing with mixed data types (like comparing a set against a list or a generator). 
# It saves you from writing extra type conversions.

def common_ids(arr_1: list, arr_2 : list ) -> list :
    set_1 = set(arr_1)
    set_2 = set(arr_2)

    common_ids = set_1 & set_2

    return list(common_ids)

system_a = [101, 102, 103, 105, 108]  
system_b = [102, 103, 104, 108, 110]

common_ids(system_a, system_b)

# THere are other set operations such as
# Union - set_a | set_b = item from both sets
# Use in with sets for speed.
# Use list(set(obj)) for de-duplication.
# Use methods (.union()) when working with lists/tuples, and operators (|) only when both sides are already strictly sets.
# Use set comprehensions {x for x in y} to clean and filter data on the fly.
````

````python
# Approach
# I can preserve order by inserting in list as traverse and check using set

def remove_dups_preserve_order(arr:list) -> list:
    check=set()
    unique_items = []
    for item in arr:
        if item not in check:
            check.add(item)
            unique_items.append(item)

    return unique_items

product_ids = [101, 103, 101, 105, 103, 108, 105]

remove_dups_preserve_order(product_ids)
````

### Q17 — Two Sum

You receive transaction amounts and a target amount.

Return:  
Indices of two transactions whose values add up to the target.

Input:  
transactions = [200, 150, 80, 120, 70]
target = 200

````python
# Approach
# Here we asked to return indices not the item related thing. so we will access item using index operator and i,e, [brackets]
# Simply to work with indices we can use enumerate func here.
# THink about make one thing to search as question is asking for 2 indeces
# but if we use 2 then it will require nested loop and make it nsquare space complexity.
# what is that one thing that can give me solution. Here is the point to think about math shorthands 

# 

def two_sum(arr:list, n:int) ->list:
    target = n
    seen = {}
    for index, item in enumerate(arr):
        comp = target - item
        if comp in seen:
            return [seen[comp], index]
        seen[item] = index

    return []

transactions = [200, 150, 80, 120, 70]
target = 200

two_sum(transactions, target)
````
