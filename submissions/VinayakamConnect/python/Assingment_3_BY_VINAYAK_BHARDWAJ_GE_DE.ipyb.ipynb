#### Q19. Data Pipeline Monitoring Your team monitors a batch ingestion pipeline. 
###### The pipeline receives these metrics:
+ total_records = 10000 
+ processed_records = 8500 
+ failed_records = 1500 
##### Write a program to: 
+ 1. Calculate the processing percentage. 
+ 2. Calculate the failure percentage. 
+ 3. Classify pipeline execution: Processing percentage 95 or above 80–94.99 Below 80 Status Healthy Warning Critical 
+ 4. Print a pipeline monitoring report. 

```python
total_records = 10000
processed_records = 8500
failed_records = 1500

processing_percent = (processed_records/total_records)*100
failure_percent= (failed_records/total_records)*100

if processing_percent >= 95:
    status = "Healthy"
elif processing_percent >= 80:
    status = "Warning"
else:
    status = "Critical"
print(f"Total Records:         {total_records}")
print(f"Processed Records:     {processed_records} ({processing_percent:.2f})")
print(f"Failed Records:        {failed_records} ({failure_percent:.2f})")
print("pipeline Status :",status)
```

#### Q20. Mini Ingestion Pipeline Capstone question: 
You are building a simple Python-based ingestion process for an e-commerce platform.
##### The source provides the following records:
+ orders = [ {"order_id": 1, "amount": 500, "status": "SUCCESS"},
+ {"order_id": 2, "amount": 1200, "status": "FAILED"}, 
+ {"order_id": 3, "amount": 800, "status": "SUCCESS"}, 
+ {"order_id": 4, "amount": -200, "status": "SUCCESS"}, {"order_id": 5, "amount": 1500, "status": "SUCCESS"} ] 
#####
 Build a program that performs the following operations: 
 + Stage 1 — Read source records Iterate through every order. 
 + Stage 2 — Validate records An order is valid only if: ● order_id is available. ● amount is greater than zero. ● status is SUCCESS. Stage 3 — Process valid records ● Collect valid orders in a separate list. ● Calculate the total revenue from valid orders. ● Count the number of valid orders. Stage 4 — Handle invalid records ● Collect invalid orders separately. ● Count the number of rejected orders. Stage 5 — Generate ingestion summary Print:  Python ========== INGESTION REPORT ========== Total records received: 5 Valid records: 3 Rejected records: 2 Total revenue: 2800 ====================================== 

```python
orders = [
    {"order_id": 1, "amount": 500, "status": "SUCCESS"},
    {"order_id": 2, "amount": 1200, "status": "FAILED"},
    {"order_id": 3, "amount": 800, "status": "SUCCESS"},
    {"order_id": 4, "amount": -200, "status": "SUCCESS"},
    {"order_id": 5, "amount": 1500, "status": "SUCCESS"}
]
valid_orders = []
invalid_orders = []
total_revenue = 0
valid_count =0
invalid_count =0 
total_count = 0
for i in orders:
    if ((i["order_id"] != None) and (i["amount"]>0))and (i["status"]=="SUCCESS"):
        valid_orders.append(i)
        total_revenue += i["amount"]
        valid_count += 1
    else :
        invalid_orders.append(i)
        invalid_count +=1 
    total_count += 1
print("========== INGESTION REPORT ========== ")
print("Total records received :",total_count)
print("Valid records :",valid_count)
print("Rejected records :",invalid_count)
print("Total revenue :",total_revenue)
print("====================================== ")


```
